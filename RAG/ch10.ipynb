## SPlitter 텍스트 분할

### 1. 문자 단위로 분할

```python
!uv add langchain-text-splitters
```

```python
with open("appendix-keywords.txt", encoding="utf-8") as f:
    file = f.read()

print (file[:500])
```

```python
#텍스트 청크 분할 방식 설정
from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    separator="\n\n", #텍스트를 구분할 구분자
    chunk_size = 210,
    chunk_overlap = 0,
    length_function = len #텍스트 길이를 계산
)
#file 텍스트를 문서 단위로 분할, list 형태로 저장
texts = text_splitter.create_documents([file]) 
print(len(texts[0].page_content)) # 글자 개수 출력
print(texts[0]) #분할 된 첫번째 문서
```

```python
#텍스트 분할에 각 청크에 메타데이터를 함께 전달
#문서 두개에 대한 메타데이터를 만듬
metadatas = [
    {"document":1},
    {"document":2},
]

#문서 두개 준비
#텍스트를 청크로 나누고, 각 청크에 해당하는 문서의 metadata를 붙여줌
documents = text_splitter.create_documents(
    [
        file,
        file,
    ],
    metadatas=metadatas,
)

print(documents[0])
print("-----")
print(len(documents))
print("-----")
print(documents[32])
```
