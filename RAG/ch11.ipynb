```python
text = ("임베딩 테슽를 하기 위한 문장")
```

### 쿼리 임베딩
embeddings.embed_query(text)는 주어진 텍스트를 임베딩 벡터로 변환
embed_query 변수에 저장

```python
#텍스트를 임베딩하여 쿼리 결과를 생성
query_result = embeddings.embed_query(text)

query_result[:5]
```

```python
# 도큐먼트 임베딩

doc_result = embeddings.embed_documents(
    [text]
)

#첫 번째 요소에서 처음 5개 항목 선택
doc_result[0][:5]
```

```python
# 차원 지정

len(doc_result[0])
```

```python
# dimensions=1024를 전달함으로써 임베딩의 크기를 1024로 줄임
# 텍스트 데이터를 1024차원의 벡터로 변환
embeddings_1024 = OpenAIEmbeddings(model = "text-embedding-3-large", dimensions=1024)

#텍스트에 대한 임베딩 벡터 생성
#리스트 첫번째요소 0은 text에 대한 임베딩 벡터
len(embeddings_1024.embed_documents([text])[0])
```

### 유사도 계산
- 임베딩한 벡테값의 유사도 계산 활용
-1에서 1사이로, 1에 가까울 수록 두 벡터가 유사 

```python
sentence1 = "안녕하세요? 반갑습니다."
sentence2 = "안녕하세요? 반갑습니다!"
sentence3 = "안녕하세요? 만나서 반가워요."
sentence4 = "Hi, nice to meet you."
sentence5 = "I like to eat apples."
```

```python
#!uv add scikit-learn
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
sentences = [sentence1, sentence2, sentence3, sentence4, sentence5]
#여러 문장들을 한 번에 임베딩
embedded_sentences = embeddings_1024.embed_documents(sentences)

# a,b의 코사인 유사도를 계산
def similarity(a,b):
    return cosine_similarity([a],[b])[0][0]

for i, sentence in enumerate(embedded_sentences):
    for j, other_sentence in enumerate(embedded_sentences):
        if i < j:
            print(
                f"[유사도 {similarity(sentence, other_sentence):.4f}]{sentences[i]}\t <====> \t {sentences[j]}"
            )
```

### 2. CacheBackedEmbeddings
- 임베딩을 한 번 수행 한 후, 그 결과를 파일 형태로 저장
- 이를 캐시 파일
- 동일한 문서에 대해 닷 임베딩을 시도하면, 먼저 캐시파일에 저장된 임베딩 값이 있는지 확인
- 캐시 파일에 값이 존재하면 추가 임베딩 작업 안해도 됨.

#### 캐시 갑 생성 방식
1) 임베딩 결과를 파일로 저장해 영구적으로 보관
2) 인메모리 바이트 스토어를 활용한 비영구적 방식, 메모리에 임시 저장되며 프로그램 종료시 데이터 삭제

```python
#vectorstore에서 임베딩 사용하기 위해
#%pip install --upgrade --quiet  langchain-openai faiss-cpu
```

```python
# 임베딩 결과를 캐시에 저장
from langchain_classic.embeddings import CacheBackedEmbeddings
# 캐시 데이터를 로컬 컴퓨터에 저장하기 위한 저장소
from langchain_classic.storage import LocalFileStore
#텍스트 파일을 읽어 도큐먼트 형태로 가져옴
from langchain_community.document_loaders import TextLoader
# 임베딩한 벡터를 저장하고 유사한 벡터를 검색하는 벡터 저장소
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings
from langchain_text_splitters import CharacterTextSplitter
```

```python
#open ai 임베딩을 사용하여 기본 임베딩 설정
embedding = OpenAIEmbeddings()

#로컬 파일 저장소 설정
store = LocalFileStore("./cache/")

#캐시를 지원하는 임베딩 생성
cached_embedder = CacheBackedEmbeddings.from_bytes_store(
    embedding,
    store,
    namespace=embedding.model, #기본 임베딩과 저장소를 사용하여 캐시 지원 임베딩을 생성
)

#store에서 키들을 순차적으로 가져옴
list(store.yield_keys())
```

```python
#문서 로드
raw_documents = TextLoader(
    "appendix-keywords.txt",
    encoding="utf-8"
).load()
#문자 단위로 텍스트 분할 설정
text_splitter = CharacterTextSplitter(chunk_size=1000, chunk_overlap = 0)
#문서 분할
documents = text_splitter.split_documents(raw_documents)
```

```python
!uv add faiss-cpu
```

```python
# 코드 실행 시간을 측정합니다.
%time db = FAISS.from_documents(documents, cached_embedder)  # 문서로부터 FAISS 데이터베이스 생성
```

```python
# 캐싱된 임베딩을 사용하여 FAISS 데이터베이스 생성
%time db2 = FAISS.from_documents(documents, cached_embedder)
```
