## CSV 로더와 데이터 프레임 로더
- CSVLoader() 사용


```python
from langchain_community.document_loaders.csv_loader import CSVLoader

FILE_PATH = "titanic.csv"
```

```python
loader = CSVLoader(FILE_PATH)

docs = loader.load()

print(len(docs))
print("\n")
print(docs[0].metadata)
```

```python
print("1번 승객 정보 : \n", docs[0].page_content)
print("\n")
print("2번 승객 정보 : \n", docs[1].page_content)
```

## CSV 내용 수정
- csv_args()를 이용하여 내용 수정

```python
loader = CSVLoader(
    file_path = "titanic.csv",
    csv_args={
        "delimiter": ",",  # 구분자
        "quotechar": '"',  # 인용 부호 문자

        #csv 파일의 각 열에 해당하는 이름
        "fieldnames": [
            "Passenger ID",
            "Survival (1: Survived, 0: Died)",
            "Passenger Class",
            "Name",
            "Sex",
            "Age",
            "Number of Siblings/Spouses Aboard",
            "Number of Parents/Children Aboard",
            "Ticket Number",
            "Fare",
            "Cabin",
            "Port of Embarkation",
        ],  # 필드 이름
    },
)

docs = loader.load()
print(docs[1].page_content)
```
