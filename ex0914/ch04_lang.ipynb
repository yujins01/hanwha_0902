```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain.agents import create_agent

#1 tool정의
def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!" #llm이 문장을 보고 city인자인 샌프란시스코를 채워서 답변함

# 2. 모델 및 에이전트 연결
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
#ai에게 너는 gpt모델을 사용하고 필요하면 get_weather라는 도구를 이용할 수 있음
agent = create_agent(model=llm, tools=[get_weather])

# 3. 실행
#사용자가 샌프란시스코 날씨어때?물으면
result = agent.invoke({"messages": [("user", "San Francisco 날씨 어때?")]})
#에이전트가 상황을 보고 날씨를 알아야함 -> get_weather tool -> 날씨 좋음 최종답변
print(result["messages"][-1].content)

# 랭스미스 Tracing에서 확인하기
```

```python
from langchain_openai import ChatOpenAI
from langchain_teddynote.models import MultiModal
from langchain_teddynote.messages import stream_response

llm2 = ChatOpenAI(
    temperature=0.1,
    model="gpt-4o-mini"
)

multiMmdal_llm = MultiModal(llm2)

IMAGE_URL = "https://upload.wikimedia.org/wikipedia/commons/3/3a/Cat03.jpg"

answer = multiMmdal_llm.stream(IMAGE_URL)

stream_response(answer)
```
