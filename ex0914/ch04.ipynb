```python
from langchain_openai import ChatOpenAI
```

```python
#챗모델 객체 생성

llm = ChatOpenAI(
    temperature= 0.1,
    model_name= "gpt-4o-mini"
)

question = "대한민국 수도는 어디?"

response = llm.invoke(question)
```

```python
print(f"[답변]: {response}")
```

## logprob 모델 답변의 확신도 
- 모델이 단어(토큰)를 생성할 때 "이 단어를 얼마나 확신을 가지고 선택했는지"에 대한 확률값을 얻을 수 있다.

```python
#logprob 모델 답변의 확신도 

#객체 생성
llm_with_logprob = ChatOpenAI(
    temperature=0.1, #창의성
    max_tokens=2048, #최대 토큰수
    model_name="gpt-4o-mini"
).bind(logprobs=True)
```

```python
question2 = "미국의 수도는?"

response2 = llm_with_logprob.invoke(question2)
```

```python
response2.response_metadata
```

```python
response2.response_metadata["logprobs"]
```
